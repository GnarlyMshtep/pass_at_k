```python
import json
import matplotlib.pyplot as plt

def plot_successful_attempts_histogram(base_path):
    def count_successful_attempts(file_path):
        uid_attempts = {}

        with open(file_path, 'r') as file:
            for line in file:
                entry = json.loads(line)
                uid = entry['uid']
                attempt_id = entry['attempt_id']
                is_correct = entry.get('is_correct', 0)

                if uid not in uid_attempts:
                    uid_attempts[uid] = set()

                if is_correct == 1:
                    uid_attempts[uid].add(attempt_id)

        # Count how many attempt ids got the question right at least once for each uid
        uid_success_counts = {uid: len(attempts) for uid, attempts in uid_attempts.items()}
        return uid_success_counts

    for i in range(50, 401, 50):
        file_path = f"{base_path}/{i}.jsonl"
        uid_success_counts = count_successful_attempts(file_path)
        print(f"Success counts for {i}: {uid_success_counts}")

        # Extract the success counts
        success_counts = list(uid_success_counts.values())

        # Filter success counts between 0 to 4
        filtered_counts = [count for count in success_counts if 0 <= count <= 4]

        # Plot the histogram
        plt.hist(filtered_counts, bins=range(6), align='left', rwidth=0.8)
        plt.xlabel('Number of Successes')
        plt.ylabel('Number of Questions')
        plt.title(f'Histogram of Number of Questions per Success Count (0 to 4) for {i}')
        plt.xticks(range(5))
        plt.show()

# Example usage
base_path = "/users/stalaei/code/pass_at_k/rollouts/full_countdown/7b_countdown_3to6_4_attempt_4_samples_1024_20250914_230923/train"
plot_successful_attempts_histogram(base_path)

```

```python
non_iid_data = "/users/stalaei/code/pass_at_k/rollouts/full_countdown/7b_countdown_3to6_4_attempt_4_samples_1024_20250914_230923/train/400.jsonl"
grpo_data = "/users/stalaei/code/pass_at_k/rollouts/full_countdown/7b_countdown_3to6_grpo_1024_20250914_231558/train/400.jsonl"

import json
from collections import defaultdict
import matplotlib.pyplot as plt

def load_entries(file_path):
    entries = {}
    with open(file_path, 'r') as f:
        for line in f:
            entry = json.loads(line)
            nums = tuple(entry['nums'])
            gts = tuple(entry['gts'])
            is_correct = entry.get('is_correct', 0)
            key = (nums, gts)
            if key not in entries:
                entries[key] = False
            if is_correct == 1:
                entries[key] = True
    return entries

def categorize_entries(non_iid_entries, grpo_entries):
    categories = defaultdict(int)
    for key in non_iid_entries.keys():
        non_iid_solved = non_iid_entries[key]
        grpo_solved = grpo_entries.get(key, False)
        
        if non_iid_solved and grpo_solved:
            categories['Both'] += 1
        elif non_iid_solved:
            categories['Non-IID Only'] += 1
        elif grpo_solved:
            categories['GRPO Only'] += 1
        else:
            categories['None'] += 1
    return categories

# Load entries from both files
non_iid_entries = load_entries(non_iid_data)
grpo_entries = load_entries(grpo_data)

# Categorize entries
categories = categorize_entries(non_iid_entries, grpo_entries)

# Plot the bar plot
plt.bar(categories.keys(), categories.values())
plt.xlabel('Category')
plt.ylabel('Number of Entries')
plt.title('Entry Success Categorization')
plt.show()
```

```python
i=2

system_prompt = (
    "You are a helpful assistant. You will be given a task to guess a hidden number. "
    "You will be asked to answer the question a couple of times, and you will win if you guess "
    "the number correctly in at least one of your attempts. Pay attention to your attempt number, "
    "and the total number of attempts you have. What is your attempt number? Think step by step and "
    "provide your answer inside <answer> YOUR GUESS HERE </answer> tags as a single integer. "
    "Don't output any other text after the </answer>."
)

# Sample 10 times and check if 'i' is in the output, for i in 1 to 4
for i in range(1, 5):
    count = 0
    for _ in range(10):
        user_prompt = (
            f"I'm thinking of a secret number between 1 and 4 (inclusive). Can you guess what it is? "
            "Please think step by step and provide your final answer inside <answer> YOUR GUESS HERE </answer> "
            f"tags as a single integer."
        )
        output = generate_text(model, 
                               prompt=user_prompt, 
                               system_message=system_prompt,
                               temperature=0.2)
        print(i, output)
        if str(i) in output:
            count += 1
    print(f"Attempt {i}: {count} out of 10 times")
```

```python
msgs = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt},
]
tokenizer = model.tokenizer
text = tokenizer.apply_chat_template(
    msgs, tokenize=False, add_generation_prompt=True
)
encoded_text = tokenizer.encode(text)
token_count = len(encoded_text)
encoded_text, token_count
```
